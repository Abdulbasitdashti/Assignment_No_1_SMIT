### 43. Ek simple calculator banayein (+, -, *, /) symbols ke basis par operation perform karein.

```python
num1 = float(input("Enter first number: "))
operator = input("Enter operation (+, -, *, /): ")
num2 = float(input("Enter second number : "))

num1 = int(num1)
num2 = int(num2)

if operator == "+":
    result = num1 + num2
    print("Result:", result)

elif operator == "-":
    result = num1 - num2
    print("Result:", result)

elif operator == "*":
    result = num1 * num2
    print("Result:", result)

elif operator == "/":
    if num2 != 0:
        result = num1 / num2
        print("Result:", result)
    else:
        print("Error: not divisible by Zero")

else:
    print("Invalid operator!")
```

### 44. User ka weight aur height le kar BMI category batayein (Underweight, Normal, Overweight).

```python
weight = float(input("Enter your weight (Kg) "))
height_cm = float(input("Enter your height (cm): "))

height = height_cm / 100
bmi = weight / (height ** 2)

print("Your BMI is:", round(bmi, 2))

if bmi < 18.5:
    print("Category: Underweight")

elif 18.5 <= bmi < 25:
    print("Category: Normal")

elif 25 <= bmi < 30:
    print("Category: Overweight")

else:
    print("Category: Obese")
```

### 58. Electricity Bill: Pehle 100 units free, next 100 units Rs 5/unit, uske baad Rs 10/unit.

```python
units = float(input("Enter Total Units: "))

bill = 0

if units <= 100:
    bill = 0
    print("First 100 units free")

elif units <= 200:
    bill = (units - 100) * 5

else:
    bill = (100 * 5) + (units - 200) * 10

print("Total Electricity Bill:", bill)
```

### 60. Final Challenge: Ek mini-project logic banayein jisme user se Username, Password, aur ek Security Question poochein nested loops aur if-else ke saath.
